```python
def prepare_company_records(companies):
    records = []

    for _, row in companies.iterrows():
        record = (
            int(row['companyId']),
            row['companyName'],
            None if pd.isna(row['AggregateRating']) else float(row['AggregateRating']),
            None if pd.isna(row['ReviewsCount']) else int(row['ReviewsCount'])
        )

        records.append(record)

    return records
```

```python
records = prepare_company_records(companies)

print(len(records))
print(records[:3])
```

```python
from psycopg2.extras import execute_values

def bulk_insert_companies(records, cursor, conn, batch_size=1000):
  for i in range (0, len(records), batch_size):
    batch = records[i:i+batch_size]
    execute_values(
        cursor,
        """
        INSERT INTO "Companies"
        (
            source_company_id,
            company_name,
            aggregate_rating,
            reviews_count
        )
        VALUES %s
        """,
        batch
    )

    print(f"Inserted {min(i + batch_size, len(records))} companies")

    conn.commit()
    print("Finished inserting all companies.")
```

```python
from psycopg2.extras import execute_values

def bulk_insert_jobs(records, cursor, conn, batch_size = 5000):
  for i in range(0, len(records), batch_size):
    batch = records[i:i+batch_size]
    execute_values(
        cursor,
        """
        INSERT INTO "Jobs"
        (
            job_title,
            raw_location,
            normalized_location,
            min_salary,
            max_salary,
            min_experience,
            max_experience,
            job_uploaded,
            description,
            company_id
        )
        VALUES %s
        """,
        batch
    )

    print(f"Inserted {min(i + batch_size, len(records))} jobs")

    conn.commit()
  print("Finished inserting all jobs.")
```

#JobSkills Table

```python
cursor.execute("""
SELECT source_job_id, job_id
FROM "Jobs"
""")

job_map = dict(cursor.fetchall())

print(len(job_map))
print(list(job_map.items())[:5])
```

```python
cursor.execute("""
SELECT skill_name, skill_id
FROM "Skills"
""")

skill_map = dict(cursor.fetchall())

print(len(skill_map))
print(list(skill_map.items())[:5])
```

```python
records = []
missing_skills = set()

for _, row in df.iterrows():

    db_job_id = job_map[row['jobId']]

    if pd.isna(row['tagsAndSkills']):
        continue

    skills = [
        skill.strip().lower()
        for skill in row['tagsAndSkills'].split(',')
    ]

    for skill in skills:
        if skill in skill_map:
          records.append(
              (
                  db_job_id,
                  skill_map[skill]
              )
          )
        else:
          missing_skills.add(skill)
```

```python
print(len(records))
print(records[:10])
print(len(missing_skills))
```

```python
records = list(dict.fromkeys(records))
print(len(records))
```

```python
from psycopg2.extras import execute_values

def bulk_insert_jobskills(records, cursor, conn, batch_size = 5000):
  for i in range(0, len(records), batch_size):
    batch = records[i:i+batch_size]
    execute_values(
        cursor,
        """
        INSERT INTO "JobSkills"
        (
            job_id,
            skill_id
        )
        VALUES %s
        """,
        batch
    )

    print(f"Inserted {min(i + batch_size, len(records))} job/skills")

    conn.commit()
  print("Finished inserting all jobs and skills ids.")
```

```python
bulk_insert_jobskills(records, cursor, conn)
```
