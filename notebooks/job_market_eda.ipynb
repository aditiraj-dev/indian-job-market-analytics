## Skills:

```python
skills = df['tagsAndSkills'].dropna().str.split(',').explode().str.strip()
print("Total skill mentions:", len(skills))
print("\nTop 20 skills:")
print(skills.value_counts().head(20))
```

```python
print("\nUnique skills:", skills.nunique())
#checking capitalization issues
print("\nSample - same skill different cases:")
print(skills[skills.str.lower() == 'python'].value_counts())
```

## Locations:

```python
print("Unique locations:", df['location'].nunique())
print("\nTop 20 locations:")
print(df['location'].value_counts().head(20))
```

```python
remote = df[df['location'].str.contains('Remote', case=False, na=False)]
hybrid = df[df['location'].str.contains('Hybrid', case=False, na=False)]
multi = df[df['location'].str.contains(',', na=False)]
print(f"\nRemote jobs: {len(remote)}")
print(f"Hybrid jobs: {len(hybrid)}")
print(f"Multi-city jobs: {len(multi)}")
```

```python
df['location'].isna().any()
```

# Salary:

```python
total = len(df)
disclosed = (df['minimumSalary'] != 0).sum()
print(f"Disclosed: {disclosed} ({round(disclosed/total*100, 1)}%)")
print(f"Undisclosed: {total - disclosed} ({round((total-disclosed)/total*100, 1)}%)")
```

```python
#distribution of disclosed salaries
disclosed_df = df[df['minimumSalary'] != 0]
print("Salary distribution of disclosed:")
print(disclosed_df['minimumSalary'].describe())
```

```python
print(df['maximumSalary'].max())
print(df['minimumSalary'].max())
```

```python
#check outliers
print("Top 10 highest minimum salaries:")
print(disclosed_df['minimumSalary'].sort_values(ascending=False).head(10))
```
